```ocaml
let id_42 n = if n <> 42 then raise (Failure "Sorry") else n;;

id_42 42;;

try id_42 0 with Failure _ -> 0;;
id_42 0;;
```

```ocaml
let id_42_res n = if n <> 42 then Error "Sorry" else Ok n;;
id_42_res 42;;
id_42_res 0;;

match id_42_res 0 with
| Ok n -> n
| Error _ -> 0;;
```
